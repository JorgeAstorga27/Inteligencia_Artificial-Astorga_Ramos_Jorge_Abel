```python
# Importar librerías necesarias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

print("Librerías cargadas exitosamente.")
```

```python
# Generar datos simulados para el ejercicio
# Simularemos 300 clientes distribuidos en zonas comerciales
X, y_verdadero = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)

# Convertir a un DataFrame de Pandas para un manejo más limpio
df = pd.DataFrame(X, columns=['Ingresos_Anuales_k$', 'Puntuacion_Gastos'])

# Mostrar las primeras filas
df.head()
```

```python
# Graficar los datos sin etiquetar para observar la distribución
plt.figure(figsize=(8, 6))
plt.scatter(df['Ingresos_Anuales_k$'], df['Puntuacion_Gastos'], s=50, color='gray', alpha=0.7)
plt.title('Distribución Inicial de Clientes (Sin Agrupar)')
plt.xlabel('Ingresos Anuales (miles de $)')
plt.ylabel('Puntuación de Gastos (1-100)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Lista para almacenar los valores de inercia (WCSS - Within-Cluster Sum of Squares)
inercia = []

# Probar valores de K desde 1 hasta 10
valores_k = range(1, 11)

for k in valores_k:
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(df)
    inercia.append(kmeans.inertia_)

# Graficar el método del codo
plt.figure(figsize=(8, 5))
plt.plot(valores_k, inercia, marker='o', linestyle='-', color='b')
plt.title('Método del Codo para Identificar el K Óptimo')
plt.xlabel('Número de Clusters (K)')
plt.ylabel('Inercia (WCSS)')
plt.xticks(valores_k)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# TODO: Asigna aquí el número de clusters que identificaste en el método del codo
k_optimo = 4 

# Inicializar y entrenar el modelo final de K-Means
kmeans_final = KMeans(n_clusters=k_optimo, init='k-means++', random_state=42, n_init=10)
df['Cluster'] = kmeans_final.fit_predict(df)

# Obtener las coordenadas de los centroides
centroides = kmeans_final.cluster_centers_

print("Modelo entrenado. Clientes asignados a sus respectivos clusters.")
df.head()
```

```python
# Graficar los clusters obtenidos y sus centroides
plt.figure(figsize=(9, 7))

# Colores para los clusters
colores = ['teal', 'orange', 'green', 'purple', 'red']

for i in range(k_optimo):
    componente_cluster = df[df['Cluster'] == i]
    plt.scatter(componente_cluster['Ingresos_Anuales_k$'], 
                componente_cluster['Puntuacion_Gastos'], 
                s=50, label=f'Cluster {i}', alpha=0.8)

# Graficar los centroides en negro con una 'X'
plt.scatter(centroides[:, 0], centroides[:, 1], s=250, c='black', marker='X', label='Centroides')

plt.title(f'Segmentación de Clientes con K-Means (K = {k_optimo})')
plt.xlabel('Ingresos Anuales (miles de $)')
plt.ylabel('Puntuación de Gastos (1-100)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```
